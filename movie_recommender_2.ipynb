```python
# ==========================================
# ADVANCED MOVIE RECOMMENDER SYSTEM
# Collaborative Filtering using MovieLens
# Downloads dataset automatically
# ==========================================

import pandas as pd
import numpy as np
import zipfile
import urllib.request
from sklearn.metrics.pairwise import cosine_similarity

url = "https://files.grouplens.org/datasets/movielens/ml-latest-small.zip"

urllib.request.urlretrieve(
    url,
    "movielens.zip"
)

with zipfile.ZipFile(
    "movielens.zip",
    'r'
) as zip_ref:
    zip_ref.extractall()

# ==========================================
# LOAD DATA
# ==========================================

movies = pd.read_csv(
    "ml-latest-small/movies.csv"
)

ratings = pd.read_csv(
    "ml-latest-small/ratings.csv"
)

# ==========================================
# CREATE USER-MOVIE MATRIX
# ==========================================

user_movie_matrix = ratings.pivot_table(
    index='movieId',
    columns='userId',
    values='rating'
).fillna(0)

print(
    "\nUser-Movie Matrix Shape:",
    user_movie_matrix.shape
)

# ==========================================
# CALCULATE MOVIE SIMILARITY
# ==========================================

print("\nCalculating Similarity Matrix...")

movie_similarity = cosine_similarity(
    user_movie_matrix
)

movie_similarity_df = pd.DataFrame(
    movie_similarity,
    index=user_movie_matrix.index,
    columns=user_movie_matrix.index
)

print("Similarity Matrix Created!")

# ==========================================
# RECOMMENDATION FUNCTION
# ==========================================

def recommend(movie_name, top_n=10):

    matches = movies[
        movies['title'].str.contains(
            movie_name,
            case=False,
            na=False
        )
    ]

    if matches.empty:
        print("\nMovie not found!")
        return

    selected_movie = matches.iloc[0]

    movie_id = selected_movie['movieId']

    print("\nSelected Movie:")
    print(selected_movie['title'])

    if movie_id not in movie_similarity_df.index:
        print("No recommendations available.")
        return

    similar_movies = movie_similarity_df[
        movie_id
    ].sort_values(
        ascending=False
    )[1:top_n+1]

    recommendations = []

    for mid in similar_movies.index:

        movie_title = movies[
            movies['movieId'] == mid
        ]['title'].values

        if len(movie_title) > 0:

            recommendations.append(
                movie_title[0]
            )

    print("\nRecommended Movies:\n")

    for i, movie in enumerate(
        recommendations,
        start=1
    ):
        print(
            f"{i}. {movie}"
        )

# ==========================================
# PROJECT STATISTICS
# ==========================================

print("\n==========================")
print("Total Movies :", len(movies))
print("Total Ratings :", len(ratings))
print("Unique Users :", ratings['userId'].nunique())
print("==========================")

# ==========================================
# USER INPUT
# ==========================================

movie_name = input(
    "\nEnter Movie Name: "
)

recommend(movie_name)
```
